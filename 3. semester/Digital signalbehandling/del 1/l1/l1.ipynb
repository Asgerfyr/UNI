1.1
Lav i Matlab en sampletidsakse. Længden skal være 0,5 sekunder, og samplingstiden Ts sættes til 0,0025
sekunder. Hvad er samplingsfrekvensen, og hvor mange samples bliver de

```python
t_0 = 0
t_slut = 0.5
T_s = 0.0025



t = np.arange(t_0, t_slut, T_s)

f_s = 1/T_s

print(f"Samplingsfrekvensen: {f_s}")

print(f"antal samples {len(t)}")
```

1.2 
Generer i Matlab et sinussignal kaldet s1(n) med f0 = 350Hz. Varigheden af sinus’en sætter vi til 1 sekund.
Benyt en samplefrekvens på 1kHz. Hvor mange samples bliver der? Gentag øvelsen - nu med en sample-
frekvens på 4kHz. Forklar forskellen mellem de to repræsentationer af sinus’e

```python
f_s1 = 350
t_s1 = 1/f_s1

t_s1_slut = 1

f_s_1 = 1000
T_s_1 = 1/f_s_1

n_samples_1 = t_s1_slut/T_s_1

print(f"Antal samples: {n_samples_1}")

f_s_2 = 4000
T_s_2 = 1/f_s_2
n_samples_2 = t_s1_slut/T_s_2
print(f"Antal samples: {n_samples_2}")


x_s1 = np.linspace(0, t_s1_slut, 100000)
x_s1_1 = np.linspace(0, t_s1_slut, int(n_samples_1))
x_s1_2 = np.linspace(0, t_s1_slut, int(n_samples_2))

y_s1 = np.sin(x_s1*f_s1*2*np.pi)
y_s1_1 = np.sin(x_s1_1*f_s1*2*np.pi)
y_s1_2 = np.sin(x_s1_2*f_s1*2*np.pi)



plt.plot(x_s1, y_s1)
plt.step(x_s1_1, y_s1_1)
plt.step(x_s1_2, y_s1_2)

plt.xlim(0,0.005)
plt.show()


```
