```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
import scipy as sci
import pandas as pd
import soundfile as sf
from IPython.display import Audio, display
```

```python
x_1 = np.array([1,4,-2,-1,5,-5,0,2,8,-0.5])
x_2 = np.array([1,-4,2,1,-5,5,0,2,-8,-0.5])

N = len(x_1)

S_1 = np.fft.fft(x_1)
S_2 = np.fft.fft(x_2)



E_1 = np.sum( np.pow( abs(S_1),2) ) /N
E_2 = np.sum( np.pow( abs(S_2),2) ) /N

print(E_1)
print(E_2)
```

```python
x_z1 = np.array([*x_1,*np.zeros(N*9)],dtype=complex)


import matplotlib.pyplot as plt

plt.plot(x_z1)
plt.xlabel('xlabel')
plt.ylabel('ylabel')
plt.title('title')
plt.grid(True)
plt.show()



S_z1 = np.fft.fft(x_z1)

plt.plot(np.unwrap(np.angle(S_z1)))
plt.xlabel('xlabel')
plt.ylabel('ylabel')
plt.title('unwrap z')
plt.grid(True)
plt.show()

plt.plot(np.unwrap(np.angle(S_1)))
plt.xlabel('xlabel')
plt.ylabel('ylabel')
plt.title('unwrap x')
plt.grid(True)
plt.show()

t_z = np.arange(0,len(S_z1)//2)
t = np.arange(0,len(S_1)//2)


plt.plot(t_z/10,20*np.log10(S_z1[:N*10//2]))
plt.plot(t,20*np.log10(S_1[:N//2]))
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.title('Frequency Spectrum')
plt.show()


plt.plot(x_1)
plt.xlabel('xlabel')
plt.ylabel('ylabel')
plt.title('title')
plt.grid(True)
plt.show()
```

```python
# Generate sine wave
f_0 = 350  # Frequency in Hz
t_duration = 1.0  # Duration in seconds
f_s = 1000  # Sampling frequency
T_s = 1/f_s

t = np.arange(0, t_duration, T_s)
x = np.sin(f_0 * 2 * np.pi * t)

plt.plot(t, x)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title(f'Sine wave {f_0} Hz')
plt.show()
```
